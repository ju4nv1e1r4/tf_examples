```python
import tensorflow as tf
import pandas as pd
```

```python
FILE_PATH = "data.csv"
```

```python
def load_csv(csv_file: str, batch_size: int, labels: list, shuffle_size: int):
    dataset = tf.data.experimental.make_csv_dataset(
        csv_file,
        batch_size=batch_size,
        num_epochs=1,
        na_value="?",
        shuffle=True,
        shuffle_buffer_size=shuffle_size
    )

    def extract_labels(features):
        extracted_labels = tuple(features.pop(label) for label in labels)
        return features, extracted_labels

    dataset = dataset.map(extract_labels)

    return dataset

# repare que adaptamos toda essa função para aceitar diversas labels
```

```python
data = load_csv(
    csv_file=FILE_PATH,
    batch_size=32,
    labels=["class_normal", "class_suspicious", "class_unknown"],
    shuffle_size=10_000
)
```
